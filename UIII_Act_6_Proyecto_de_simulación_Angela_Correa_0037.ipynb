#### Nómina electrónica con if, else, elif y while

```python
print("==================================================")
print("     BIENVENIDO A NOMINA-IA: TU ASESOR LABORAL")
print("==================================================")
print("IA: Hola, soy tu asistente de nómina electrónica.")
print("Te ayudaré a calcular tu pago neto y entender tus descuentos.")

# Entrada de datos iniciales
print("\nIA: Primero, configuremos los datos de tu contrato.")
nombre = input("¿Cuál es tu nombre completo?: ")
salario_texto = input("¿Cuál es tu sueldo bruto mensual (en tu moneda local)?: ")
horas_extra_texto = input("¿Cuántas horas extra trabajaste este mes?: ")

# Conversión manual de tipos de datos
salario_bruto = float(salario_texto)
horas_extra = int(horas_extra_texto)

print("\nIA: ¡Perfecto, " + nombre + "! Información procesada con éxito.")

# Variable de control para el bucle principal
ejecutar = "si"

# Bucle interactivo
while ejecutar == "si":
    print("\n------------------------------------------------")
    print("¿Qué deseas consultar hoy? Escribe una opción:")
    print("- liquidar : Calcula tu pago neto y deducciones.")
    print("- legal    : Te explico por qué la nómina es electrónica.")
    print("- salir    : Cerrar el asistente de nómina.")
    print("------------------------------------------------")

    opcion = input("Tu elección: ").lower()

    if opcion == "salir":
        ejecutar = "no"
        print("\nIA: Gracias por usar Nomina-IA. Tu registro ha sido enviado a la nube. ¡Adiós!")

    elif opcion == "legal":
        print("\n[IA - FUNDAMENTOS LEGALES]")
        print("La nómina electrónica es obligatoria para que las empresas puedan")
        print("deducir los costos salariales de sus impuestos.")
        print("Además, sirve como prueba legal de tus aportes a seguridad social.")

    elif opcion == "liquidar":
        print("\n[IA - CALCULANDO NÓMINA ELECTRÓNICA...]")

        # --- LÓGICA DE CÁLCULO USANDO CONDICIONALES ---

        # 1. Cálculo de Seguridad Social (Simulamos un 8% de deducción total)
        salud = salario_bruto * 0.04
        pension = salario_bruto * 0.04
        total_deducciones = salud + pension

        # 2. Cálculo de Horas Extra (Valor estimado por hora)
        valor_hora = (salario_bruto / 30) / 8
        pago_extra = horas_extra * (valor_hora * 1.25)

        # 3. Cálculo de Neto a Pagar
        neto_pagar = salario_bruto + pago_extra - total_deducciones

        print("IA: Se han aplicado deducciones de Salud (4%) y Pensión (4%).")
        print("IA: Valor de Horas Extra: " + str(round(pago_extra, 2)))
        print("IA: TOTAL NETO A RECIBIR: " + str(round(neto_pagar, 2)))

        # Evaluación de perfil salarial
        if salario_bruto < 1500:
            print("\nIA Análisis: Tu nivel salarial suele aplicar para subsidio de transporte.")
            print("Consejo: Verifica que este subsidio esté incluido en tu colilla electrónica.")
        elif salario_bruto >= 1500 and salario_bruto <= 4000:
            print("\nIA Análisis: Tienes un salario de rango medio.")
            print("Consejo: Revisa si tu empresa ofrece planes de ahorro voluntario por nómina.")
        else:
            print("\nIA Análisis: Rango salarial alto detectado.")
            print("Consejo: Es posible que se te aplique retención en la fuente adicional.")

    else:
        print("\nIA: Error. No reconozco el comando '" + opcion + "'.")
        print("Intenta con 'liquidar', 'legal' o 'salir'.")

print("\n==================================================")
print("       SIMULADOR DE NÓMINA FINALIZADO")
print("       Programado por Angela Correa NL = 14")
print("==================================================")
```
